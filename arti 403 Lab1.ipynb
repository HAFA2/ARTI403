```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from PIL import Image

# 1. التأكد من إنشاء مجلد images والصور المطلوبة
os.makedirs('images', exist_ok=True)

# توليد الصورتين وحفظهما محلياً
dummy_data = (np.random.rand(256, 256) * 255).astype(np.uint8)
Image.fromarray(dummy_data).save('images/cameraman.tif')
Image.fromarray(dummy_data).save('images/lena_gray_256.tif')

# Task #2: Loading and Visualizing Images
# a. Loading/reading an image using OpenCV
img = cv2.imread('images/cameraman.tif', cv2.IMREAD_UNCHANGED)
plt.figure(figsize=(3, 3))
plt.imshow(img, cmap='gray')
plt.title("Loaded via OpenCV")
plt.axis('off')
plt.show()

# b. Loading/reading image using PIL
img2 = Image.open('images/lena_gray_256.tif')
plt.figure(figsize=(3, 3))
plt.imshow(img2, cmap=cm.Greys_r)
plt.title("Loaded via PIL")
plt.axis('off')
plt.show()

# Task #3: Image Storing
cv2.imwrite('new_image.jpg', img)
img2.save('new_image2.jpg')

# Task #4: Display image as an Array
print("OpenCV Array Shape:", img.shape)
print(img)

img_array = np.array(img2)
print("\nPIL Array Shape:", img_array.shape)
print(img_array)
```
