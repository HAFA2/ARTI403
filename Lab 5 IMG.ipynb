```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
from skimage import data

img1 = data.astronaut()  
```

```python
# استخدام صورة جاهزة مدمجة
img1 = data.astronaut()

# تطبيق Box Filter بحجم 7x7
smoothed_box = cv2.blur(img1, (7, 7))

# عرض الصورتين جنباً إلى جنب
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img1)
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(smoothed_box)
axes[1].set_title('Smoothed (7x7 Box Filter)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
img2 = data.chelsea()
gaussian_5x5 = cv2.GaussianBlur(img2, (5, 5), sigmaX=0)
gaussian_21x21 = cv2.GaussianBlur(img2, (21, 21), sigmaX=0)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img2)
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(gaussian_5x5)
axes[1].set_title('Gaussian Filter (5x5)')
axes[1].axis('off')

axes[2].imshow(gaussian_21x21)
axes[2].set_title('Gaussian Filter (21x21)')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
img3 = data.astronaut()
image_float = img3.astype(np.float32) / 255.0
laplacian = cv2.Laplacian(image_float, ddepth=cv2.CV_32F, ksize=3)
sharpened = np.clip(image_float - laplacian, 0, 1)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img3)
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(np.clip(laplacian, 0, 1))
axes[1].set_title('Laplacian Filter (3x3)')
axes[1].axis('off')

axes[2].imshow(sharpened)
axes[2].set_title('Sharpened Image')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
